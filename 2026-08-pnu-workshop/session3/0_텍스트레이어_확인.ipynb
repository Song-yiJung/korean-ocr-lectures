# 자료 받기

## `!` 가 붙은 줄

`!` 는 **프로그램을 실행하라**는 표시입니다.

| | |
|---|---|
| `!pip install …` | 도구를 깝니다. `-q` 는 조용히 |
| `!wget … -O "이름"` | 인터넷에서 받아 그 이름으로 저장합니다 |

빌린 컴퓨터라 **매번 새로 깝니다.** 내 컴퓨터에는 남지 않습니다.

## `저장소 = "https://…"`

이름표를 붙입니다. 아래에서 `{저장소}` 라고 부르면 저 주소가 들어갑니다.

## `for` — 되풀이

```
for f in ("파일1", "파일2"):
```

괄호 안의 것마다 아래 줄을 되풀이합니다. **들여쓴 줄까지가 한 묶음**입니다.

`os.path.getsize(f)` 가 파일 크기를 재고 `/ 1024` 로 KB가 됩니다.

````python
!pip install -q pymupdf

저장소 = "https://raw.githubusercontent.com/Song-yiJung/korean-ocr-lectures/main"

!wget -q "{저장소}/2026-08-pnu-workshop/session3/data/drag/drag_A_busan60.pdf"  -O "긁어보기_A_부산광복60년.pdf"
!wget -q "{저장소}/2026-08-pnu-workshop/session3/data/drag/drag_B_gusulsa.pdf"  -O "긁어보기_B_부산구술사총서.pdf"

import os
for f in ("긁어보기_A_부산광복60년.pdf", "긁어보기_B_부산구술사총서.pdf"):
    크기 = os.path.getsize(f) / 1024 if os.path.exists(f) else 0
    표시 = "" if 크기 > 100 else "   ⚠ 받기 실패 — 손을 들어 주세요"
    print(f"{f:32} {크기:>7,.0f}KB{표시}")
````

# 파일 안에 글자가 들어 있나

## `import fitz`

도구를 꺼내 옵니다. 꺼내야 씁니다.
`pymupdf` 로 깔았는데 코드에서는 `fitz` 로 부릅니다. 원래 그런 도구입니다.

| | |
|---|---|
| `fitz.open(파일)` | PDF를 엽니다 |
| `쪽.get_text()` | 그 쪽에 든 글자를 꺼냅니다. 없으면 빈 값 |
| `.strip()` | 앞뒤 빈칸과 줄바꿈을 떼어냅니다 |
| `len(…)` | 개수를 셉니다. 여기서는 글자 수 |
| `문서.close()` | 닫습니다 |

## `[len(…) for 쪽 in 문서]`

대괄호 안에 `for` 가 들어가 있습니다. **모든 쪽을 돌며 글자 수를 세어 목록으로 만드는**
짧은 꼴입니다. 풀어 쓰면 이렇습니다.

```
쪽별 = []
for 쪽 in 문서:
    쪽별.append(len(쪽.get_text().strip()))
```

`sum(쪽별)` 로 합계를 내고, `sum(1 for n in 쪽별 if n)` 으로 0이 아닌 쪽을 셉니다.

````python
import fitz   # PyMuPDF — PDF 안을 들여다보는 도구

for 파일 in ("긁어보기_A_부산광복60년.pdf", "긁어보기_B_부산구술사총서.pdf"):
    문서 = fitz.open(파일)
    쪽별 = [len(쪽.get_text().strip()) for 쪽 in 문서]
    판정 = "OCR 필요" if sum(쪽별) == 0 else "OCR 불필요"
    print(f"{파일:32} {len(문서):>3}쪽   "
          f"글자 있는 쪽 {sum(1 for n in 쪽별 if n):>3}   "
          f"{sum(쪽별):>8,}자   {판정}")
    문서.close()
````

# 한 쪽만 꺼내서 확인

| | |
|---|---|
| `문서[쪽번호 - 1]` | **`- 1` 을 눈여겨보십시오.** 컴퓨터는 0부터 세서 1쪽이 0번입니다 |
| `글[:300]` | 앞에서 300자만. 다 찍으면 화면이 넘칩니다 |
| `if 글 else "(비어 있습니다)"` | 있으면 앞의 것, 없으면 뒤의 것을 찍습니다 |
| `"─" * 60` | 같은 글자를 60번 되풀이합니다. 구분선 긋기 |

> `쪽번호` 를 바꿔 가며 눌러 보셔도 됩니다.

````python
쪽번호 = 1   # ← 바꿔 가며 확인해 보셔도 됩니다

for 파일 in ("긁어보기_A_부산광복60년.pdf", "긁어보기_B_부산구술사총서.pdf"):
    문서 = fitz.open(파일)
    글 = 문서[쪽번호 - 1].get_text().strip()
    print("─" * 60)
    print(f"{파일}  {쪽번호}쪽 — {len(글):,}자")
    print("─" * 60)
    print(글[:300] if 글 else "(비어 있습니다)")
    print()
    문서.close()
````
